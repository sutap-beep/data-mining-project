### 01 BUSINESS UNDERSTANDING:
1. The industry uses their machinery continuously to manifacture output based on demand. This exposes the machinery to wear and tear caused by generated temperature, pressure upon continuous use.
2. The maintenance team strategy rely on strict time schedule maintenance and reactive maintenance unexpected breakdown on the machinery. This halt production line, delay customer order, and incure high operation cost.

### BUSINESS OBJECTIVE
1. Reduce unschedule machine downtime and lower maintenance costs.

#### BUSINESS SUCCESS CRITERIA:
1. To lower the operating cost by reducing unexpected breakdown.
2. To minimize uneccesary schedule maintenance for healthy machinery.

#### DATA MINING GOAL:
1. Build Classification model capable of detecting non-linear operational anomalies and predicting machine failure before they occur.

#### DATA MINING SUCCESS METRICS:
1. Achieve recall score 85% and above.
2. Achieve Precision score 70% and above.
