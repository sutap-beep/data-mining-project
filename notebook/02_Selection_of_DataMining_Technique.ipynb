### SELECTION OF DATA MINING TECHNIQUE

DATA MINING TECHNIQUE: CLASSIFICATION USING LIGHT GRADIENT BOOSTING MACHINE (LightGBM)
1. Classification Data Mining Technique is choosen for this dataset because it directly translate raw engineering sensor data into clear, actionable maintenance decisions.
2. Alligns with predictive maintenance goal as it answer the question as to whether the machine fail or continue to operate in the future. As classification algorithm naturally output a probability of discrete classes, 0 for 'Operate as Normal' and 1 for 'Machine Failure'.
3. Machine Failure is cause by non-linear physical interactions such as 'Heat Dissipation Failure (HDF) and 'Power Failure (PWF)', and LightBGM is a tree based classification model which excel at solving multi-dimensional boundaries across features.
4. Since the machine runs smoothly all the time predictive maintenance dataset is always imbalance. LightGBM have built-in mathematical parameters that is designed to handle class imbalances, ensuring the model learns the pattern of machine failure without being overcome by normal data.

Goal:
1. To develop classifier using LightGBM capable of identifying non-linear sensors patterns that causes breakdown.
2. Overcome the dataset's severe class imbalance.
3. Maximize Precision-Recall Area Under the Curve (PR-AUC)
